```python
import torch
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
```

check if cuda is available otherwise use cpu

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
```

Choose pretrained model "Helsinki-NLP/opus-mt-sv-en"

```python
model_name = "Helsinki-NLP/opus-mt-sv-en"

tokenizer = AutoTokenizer.from_pretrained(model_name) # down load a tokenizer for the helsinki model
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

model = model.to(device)
```

Translate function from sv to en

```python
def translate_sv_to_en(sentences: list[str]) -> list[str]:
    inputs = tokenizer(
        sentences,
        return_tensors="pt", # return pytorch tensors
        padding=True,
        truncation=True
    ).to(device)

    with torch.no_grad():
        output_ids = model.generate(
            **inputs,
            max_new_tokens=50
        )

    translation = tokenizer.decode(
        output_ids,
        skip_special_tokens=True
    )

    return translation
```

Test the model


```python
model.eval()
sentences = [
    "Jag gillar maskininlärning.",
    "Hur mår du idag?",
    "Det här är ett litet test.",
    "Jag studerar datavetenskap."
]
translations=translate_sv_to_en(sentences)
for i in translations:
    print(i)
```
